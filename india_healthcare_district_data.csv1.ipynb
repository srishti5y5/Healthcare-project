```python
import pandas as pd
import numpy as np

# Load dataset

df = pd.read_csv("C:/Users/Dell/Downloads/india_healthcare_district_data.csv")
df.head()
```

```python
df = df.drop_duplicates(subset=['district_id', 'year'])
```

```python
num_cols = [
    'population', 'urban_population', 'rural_population',
    'number_of_doctors', 'number_of_hospitals',
    'hospital_beds', 'primary_health_centers'
]

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df['year'] = df['year'].astype(int)
df['population'] = df['urban_population'] + df['rural_population']
```

```python
df = df[df['population'] > 0]
df = df[df['number_of_doctors'] >= 0]
df = df[df['hospital_beds'] >= 0]
```

```python
df['doctor_to_population_ratio'] = df['number_of_doctors'] / df['population']
df['bed_availability_index'] = df['hospital_beds'] / df['population']
```

```python
for col in ['population', 'number_of_doctors', 'hospital_beds']:
    upper_limit = df[col].quantile(0.99)
    df[col] = np.where(df[col] > upper_limit, upper_limit, df[col])
```

```python
df['district_name'] = df['district_name'].str.strip().str.title()
df['state'] = df['state'].str.strip().str.title()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['healthcare_access_score']] = scaler.fit_transform(df[['healthcare_access_score']])
```

```python
df['healthcare_pressure'] = df['population'] / (df['number_of_doctors'] + 1)

df.to_csv("clean_healthcare_data.csv", index=False)

print("✅ Done")
```
